```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# =========================
# 1. MONITORING STATIONS
# =========================
stations = [
    {"station_id": "STN-EDSA-01", "name": "EDSA-Guadalupe", "traffic": "Very High", "zone": "Commercial", "city": "Quezon City"},
    {"station_id": "STN-ESP-02", "name": "España-UST", "traffic": "High", "zone": "Urban", "city": "Manila"},
    {"station_id": "STN-C5-03", "name": "C5-Pasig", "traffic": "Mixed", "zone": "Arterial", "city": "Pasig"},
    {"station_id": "STN-QUI-04", "name": "Quiapo-Plaza Miranda", "traffic": "Very High", "zone": "Urban Core", "city": "Manila"},
    {"station_id": "STN-UPD-05", "name": "UP Diliman", "traffic": "Low", "zone": "Green Zone", "city": "Quezon City"},
]

stations_df = pd.DataFrame(stations)

# =========================
# 2. SENSORS
# =========================
sensors = []

sensor_types = ["AQM", "MET"]  # Air Quality + Meteorological

for i, station in enumerate(stations):
    for stype in sensor_types:
        sensors.append({
            "sensor_id": f"{stype}-{i+1:03d}",
            "station_id": station["station_id"],
            "sensor_type": "Air Quality" if stype == "AQM" else "Weather",
            "model": "IoT-SIM-2026",
            "status": "Active"
        })

sensors_df = pd.DataFrame(sensors)

# =========================
# 3. TRAFFIC IMPACT FUNCTION
# =========================
def traffic_multiplier(level):
    return {
        "Very High": 1.8,
        "High": 1.5,
        "Mixed": 1.2,
        "Low": 0.8
    }.get(level, 1.0)

# =========================
# 4. SENSOR READINGS (IoT DATA)
# =========================
num_records = 300
readings = []

for _ in range(num_records):
    station = np.random.choice(stations)
    multiplier = traffic_multiplier(station["traffic"])

    pm25 = np.random.uniform(5, 40) * multiplier
    pm10 = np.random.uniform(10, 80) * multiplier
    co = np.random.uniform(0.2, 2.0) * multiplier
    no2 = np.random.uniform(10, 120) * multiplier
    o3 = np.random.uniform(5, 60)

    temperature = np.random.uniform(24, 36)
    humidity = np.random.uniform(50, 90)
    wind_speed = np.random.uniform(0.5, 6.0)

    # AQI computation (simplified real-world approximation)
    aqi = (pm25 * 2.0) + (pm10 * 1.2) + (no2 * 1.5) + (co * 10)

    if aqi <= 50:
        category = "Good"
    elif aqi <= 100:
        category = "Moderate"
    elif aqi <= 150:
        category = "Unhealthy for Sensitive"
    else:
        category = "Unhealthy"

    readings.append({
        "timestamp": datetime.now() - timedelta(minutes=np.random.randint(0, 1440)),
        "station_id": station["station_id"],
        "sensor_id": f"AQM-{np.random.randint(1,6):03d}",
        "location": station["name"],
        "pm25": round(pm25, 2),
        "pm10": round(pm10, 2),
        "co": round(co, 2),
        "no2": round(no2, 2),
        "o3": round(o3, 2),
        "temperature": round(temperature, 1),
        "humidity": round(humidity, 1),
        "wind_speed": round(wind_speed, 2),
        "aqi": round(aqi, 2),
        "aqi_category": category
    })

readings_df = pd.DataFrame(readings)

# =========================
# 5. SAVE FILES (3-TABLE SYSTEM)
# =========================
stations_df.to_csv("stations.csv", index=False)
sensors_df.to_csv("sensors.csv", index=False)
readings_df.to_csv("sensor_readings.csv", index=False)

stations_df.to_json("stations.json", orient="records")
sensors_df.to_json("sensors.json", orient="records")
readings_df.to_json("sensor_readings.json", orient="records")

# =========================
# 6. DISPLAY SAMPLE
# =========================
print("STATIONS SAMPLE")
print(stations_df.head(), "\n")

print("SENSORS SAMPLE")
print(sensors_df.head(), "\n")

print("READINGS SAMPLE")
print(readings_df.head())
```
